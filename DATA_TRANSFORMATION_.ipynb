```python
q1=df['carat'].quantile(0.25)
print(q1)
```

```python
q3=df['carat'].quantile(0.75)
print(q3)
```

```python
iqr=q3-q1
print(iqr)
```

```python
upper_bound=q3+1.5*iqr
print(upper_bound)
```

```python
lower_bound=q1-1.5*iqr
print(lower_bound)
```

```python
df['carat']=df['carat'].clip(lower_bound,upper_bound)
```

```python
# to check the outliers are present
sns.boxplot(df['carat'])
```

```python
df.drop(columns="Unnamed: 0",inplace=True)
df
```

```python
# to check outlier in all the columns
for col in df.select_dtypes(exclude=['object']).columns:
  sns.boxplot(y=df[col])
  plt.title(f'box plot of {col}')
  plt.show()
```

```python
# automated code for outlier capping
for i in df.select_dtypes(exclude=['object']).columns:
  q1=df[i].quantile(0.25)
  q3=df[i].quantile(0.75)
  iqr=q3-q1
  upper_bound=q3+1.5*iqr
  lower_bound=q1-1.5*iqr
  df[i]=df[i].clip(lower_bound,upper_bound)
```

```python
# to check outlier in all the columns after the outlier capping
for col in df.select_dtypes(exclude=['object']).columns:
  sns.boxplot(y=df[col])
  plt.title(f'box plot of {col}')
  plt.show()
```
